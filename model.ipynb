```python
import torchvision
import torch.nn as nn
import torch
vgg16=torchvision.models.vgg16(progress=False)
```

```python
print(vgg16)

```

```python
vgg16.classifier[6]=nn.Linear(1000,10)
print(vgg16)
```

```python
vgg16.add_module("test",nn.Linear(1000,10))
print(vgg16)
```

```python
vgg16.classifier.add_module("test",nn.Linear(1000,10))
print(vgg16)
```

```python
torch.save(vgg16,"1.pth")
torch.load("1.pth")
#需要导入对应的类，本质和下面差不多
```

```python
torch.save(vgg16.state_dict(),"2.pth")
vgg16.load_state_dict(torch.load("2.pth"))
```
