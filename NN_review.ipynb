```python
import torch
import torchvision
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
import torch.nn as nn
import torch.nn.functional as F
```

```python
class My_NN(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.conv1=nn.Conv2d(in_channels=3,out_channels=6,kernel_size=3,stride=1,padding=0)
    def forward (self,input):
        return self.conv1(input)
    
```

```python
dataset=torchvision.datasets.CIFAR10(root="./dataset",train=False,transform=torchvision.transforms.ToTensor(),download=True)
dataloader=DataLoader(dataset=dataset,batch_size=64,shuffle=True,drop_last=False)
net=My_NN()
writer=SummaryWriter("./logs")
```

```python
step=0
for data in dataloader:
    imgs,target=data
    output=net(imgs)
    # print(imgs.shape)
    # print(output.shape)
    writer.add_images("in",imgs,step)
    writer.add_images(tag="out",img_tensor=torch.reshape(input=output,shape=(-1,3,30,30)),global_step=step)
    step+=1

writer.close()
```
