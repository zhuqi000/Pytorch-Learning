```python
import torch
import torchvision
import torch.nn as nn 
import torch.optim as optim
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
```

```python
class My_Sequential(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.conv1 = nn.Conv2d(in_channels=3,out_channels=32,kernel_size=5,padding="same")
        self.maxpool1=nn.MaxPool2d(kernel_size=2)
        self.conv2 = nn.Conv2d(in_channels=32,out_channels=32,kernel_size=5,padding="same")
        self.maxpool2=nn.MaxPool2d(kernel_size=2)
        self.conv3 = nn.Conv2d(in_channels=32,out_channels=64,kernel_size=5,padding="same")
        self.maxpool3=nn.MaxPool2d(kernel_size=2)
        self.flatten = nn.Flatten()
        self.liner1 = nn.Linear(in_features=64*4*4,out_features=64,bias=True)
        self.liner2 = nn.Linear(in_features=64,out_features=10,bias=True)

    def forward(self,input):
        output = nn.Sequential(self.conv1,self.maxpool1,
                               self.conv2,self.maxpool2,
                               self.conv3,self.maxpool3,
                               self.flatten,self.liner1,self.liner2)(input)
        return output
```

```python
net = My_Sequential()
test_input=torch.randn((64,3,32,32))
output=net(test_input)
print(output.shape)
```

```python
dataset=torchvision.datasets.CIFAR10(root="./dataset",train=True,
                                     transform=torchvision.transforms.ToTensor(),download=True)
dataload = DataLoader(dataset=dataset,batch_size=64,shuffle=False)
loss = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(),lr=0.01)
writer=SummaryWriter("./log")
step=0
for data in dataload:
    optimizer.zero_grad()
    imgs,target=data
    output=net(imgs)
    return_loss=loss(output,target)
    return_loss.backward()
    optimizer.step()
    writer.add_scalar(tag="loss",scalar_value=return_loss,global_step=step)
    step+=1
writer.close()
```
